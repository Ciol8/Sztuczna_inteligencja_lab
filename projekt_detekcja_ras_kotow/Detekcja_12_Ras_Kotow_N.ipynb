```
from roboflow import Roboflow
from ultralytics import YOLO
import os
```

```
# ==============================================================================
# KROK 2: POBRANIE IDEALNIE PODZIELONEGO ZBIORU DANYCH Z ROBOFLOW
# ==============================================================================
print("Pobieranie zbioru danych 12 ras kotów...")

# Autoryzacja i pobranie danych.
# Ten zbiór ma już wbudowany podział: 80% Train, 10% Valid, 10% Test.
# Dzięki temu trzeba ręcznie dzielić zdjęć, a mechanizmy takie jak
# Early Stopping zadziałają od razu poprawnie.
rf = Roboflow(api_key="LwahPAwwo4dka3e3rl9E")
project = rf.workspace("machine-learning-qir8h").project("cat-breed-detection-2snvs")
version = project.version(2)
dataset = version.download("yolov8")
```

```
# ==============================================================================
# KROK 3: INICJALIZACJA MODELU
# ==============================================================================
print("\nInicjalizacja modelu YOLOv8n...")

# nano
model = YOLO('yolov8n.pt')
```

```
# ==============================================================================
# KROK 4: TRENING MODELU
# ==============================================================================
print("\nRozpoczęcie treningu sieci neuronowej...")

results = model.train(
    data=f"{dataset.location}/data.yaml", # Gotowy plik konfiguracyjny pobrany z Roboflow
    epochs=60,                            # Uczymy przez 50 przejść przez zbiór danych
    imgsz=640,                            # Zdjęcia są skalowane do 640x640px
    batch=16,                             # Mini-batch: pakiet 16 zdjęć przed aktualizacją wag

    # --- HIPERPARAMETRY I REGULARYZACJA ---
    optimizer='Adam',                     # Algorytm adaptacyjny optymalizacji spadku wzdłuż gradientu
    lr0=0.001,                            # Początkowy współczynnik uczenia (Learning Rate)
    patience=10,                          # Early Stopping: przerwij jeśli val_loss nie maleje przez 10 epok
    dropout=0.1,                          # Regularyzacja: losowe wyłączanie 10% neuronów (przeciwdziała przeuczeniu)

    name="model_12_ras_kotow"             # Nazwa folderu do zapisu wag modelu
)
```

```
# ==============================================================================
# KROK 5: TESTOWANIE NA NIEWIDZIANYM ZBIORZE
# ==============================================================================
print("\nTrening zakończony! Zapisano model. Czas na inferencję (test)...")

# Model zapisuje swoje najlepsze wagi pod nazwą best.pt
najlepszy_model = YOLO('runs/detect/model_12_ras_kotow/weights/best.pt')

# Szukamy folderu "test" (który stanowi 10% całego zbioru, czyli ok. 238 zdjęć)
test_dir = os.path.join(dataset.location, "test", "images")
if os.path.exists(test_dir):
    lista_zdjec = os.listdir(test_dir)
    if len(lista_zdjec) > 0:
        # Wyciągamy pierwsze testowe zdjęcie do sprawdzenia
        for zdjecie in range(len(lista_zdjec)):

          przykladowe_zdjecie = os.path.join(test_dir, lista_zdjec[zdjecie])

          # Predykcja (conf=0.5 oznacza, że chcemy widzieć tylko pewne wyniki >50%)
          predykcja = najlepszy_model.predict(source=przykladowe_zdjecie, save=True, conf=0.5)
        print(f"\nUkończono pomyślnie! Wynik testu zapisano w folderze 'runs/detect/predict'.")
```
