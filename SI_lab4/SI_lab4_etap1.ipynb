```python
import numpy as np
from sklearn import preprocessing
raw_csv_data = np.loadtxt('Audiobooks_data.csv',delimiter=',')
unscaled_inputs_all = raw_csv_data[:,1:-1]
targets_all = raw_csv_data[:,-1]
```

```python
shuffeled_indices = np.arange(unscaled_inputs_all.shape[0])
np.random.shuffle(shuffeled_indices)
unscaled_inputs_all = unscaled_inputs_all[shuffeled_indices]
targets_all = targets_all[shuffeled_indices]
```

```python
num_one_targets = int(np.sum(targets_all))
zero_targets_counter = 0
indices_to_remove = []
for i in range(targets_all.shape[0]):
    if targets_all[i] == 0:
        zero_targets_counter += 1
        if zero_targets_counter > num_one_targets:
            indices_to_remove.append(i)
unscaled_inputs_equal_priors = np.delete(unscaled_inputs_all, indices_to_remove, axis=0)
targets_equal_priors = np.delete(targets_all, indices_to_remove, axis=0)
```

```python
scaled_inputs=preprocessing.scale(unscaled_inputs_equal_priors)
```

```python
shuffeled_indices = np.arange(scaled_inputs.shape[0])
np.random.shuffle(shuffeled_indices)
shuffeled_inputs = scaled_inputs[shuffeled_indices]
shuffeled_targets = targets_equal_priors[shuffeled_indices]
```

```python
sample_count = shuffeled_inputs.shape[0]
train_sample_count = int(0.8*sample_count)
validation_sample_count = int(0.1*sample_count)
test_sample_count = sample_count - train_sample_count - validation_sample_count
train_inputs = shuffeled_inputs[:train_sample_count]
train_targets = shuffeled_targets[:train_sample_count]
validation_inputs = shuffeled_inputs[train_sample_count:train_sample_count+validation_sample_count]
validation_targets = shuffeled_targets[train_sample_count:train_sample_count+validation_sample_count]
test_inputs = shuffeled_inputs[train_sample_count+validation_sample_count:]
test_targets = shuffeled_targets[train_sample_count+validation_sample_count:]
```

```python
print(np.sum(train_targets), train_sample_count, np.sum(train_targets) / train_sample_count)
print(np.sum(validation_targets), validation_sample_count, np.sum(validation_targets) / validation_sample_count)
print(np.sum(test_targets),test_sample_count, np.sum(test_targets) / test_sample_count)
```

```python
np.savez('Audiobooks_data_train', inputs=train_inputs, targets=train_targets)
np.savez('Audiobooks_data_validation', inputs=validation_inputs, targets=validation_targets)
np.savez('Audiobooks_data_test', inputs=test_inputs, targets=test_targets)
```
